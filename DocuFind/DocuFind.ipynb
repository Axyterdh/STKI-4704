```python
!pip install Sastrawi

import re
from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
from sklearn.feature_extraction.text import TfidfVectorizer

# Mempersiapkan objek stemmer dan stopword
stemmer_factory = StemmerFactory()
stemmer = stemmer_factory.create_stemmer()

stopword_factory = StopWordRemoverFactory()
stopwords = set(stopword_factory.get_stop_words())

def preprocessing(teks):
    # 1. Case folding
    teks = teks.lower()
    # 2. Menghapus angka dan karakter khusus
    teks = re.sub(r'[^a-zA-Z\s]', ' ', teks)
    # 3. Tokenization
    tokens = teks.split()
    # 4. Stopword removal
    tokens = [kata for kata in tokens if kata not in stopwords]
    # 5. Stemming
    tokens = [stemmer.stem(kata) for kata in tokens]
    return ' '.join(tokens)

# Lakukan preprocessing pada data yang baru dimuat
df["teks_preprocessing"] = df["isi_dokumen"].apply(preprocessing)

# Proses pembobotan TF-IDF
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(
    df["teks_preprocessing"]
)

print("Perhitungan TF-IDF berhasil!")
print("Ukuran matriks:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Menghitung kemiripan antar seluruh dokumen
similarity_matrix = cosine_similarity(tfidf_matrix)

print("Cosine Similarity berhasil dihitung!")
print("Ukuran matriks:", similarity_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def cari_dokumen(query, jumlah_hasil=5):

    # Preprocessing kata kunci pencarian
    query_preprocessing = preprocessing(query)

    # Mengubah query menjadi vektor TF-IDF
    query_vector = tfidf_vectorizer.transform([query_preprocessing])

    # Menghitung kemiripan query dengan seluruh dokumen
    skor_similarity = cosine_similarity(
        query_vector,
        tfidf_matrix
    ).flatten()

    # Mengurutkan dokumen berdasarkan skor tertinggi
    hasil_urut = skor_similarity.argsort()[::-1]

    print("Kata kunci:", query)
    print("\nHasil pencarian DocuFind:\n")

    jumlah_ditampilkan = 0

    for indeks in hasil_urut:

        # Mengabaikan dokumen dengan skor 0
        if skor_similarity[indeks] <= 0:
            continue

        print("Judul:", df.iloc[indeks]["judul"])
        print("Kategori:", df.iloc[indeks]["kategori"])
        print("Skor Similarity:", round(skor_similarity[indeks], 4))
        print("-" * 50)

        jumlah_ditampilkan += 1

        if jumlah_ditampilkan >= jumlah_hasil:
            break

    if jumlah_ditampilkan == 0:
        print("Tidak ditemukan dokumen yang relevan.")
```

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity

# Daftar query pengujian
queries = {
    "Q1": "belajar pemrograman python",
    "Q2": "belajar database dan SQL",
    "Q3": "jaringan komputer dan protokol",
    "Q4": "kecerdasan buatan",
    "Q5": "membuat dan mengembangkan website"
}

# Ground truth: dokumen yang dianggap relevan
ground_truth = {
    "Q1": {"D01", "D02", "D03", "D04"},
    "Q2": {"D05", "D06", "D07", "D08"},
    "Q3": {"D09", "D10", "D11", "D12"},
    "Q4": {"D13", "D14", "D15", "D16"},
    "Q5": {"D17", "D18", "D19", "D20"}
}

# Jumlah hasil pencarian
top_k = 5

hasil_evaluasi = []

for kode_query, query in queries.items():

    # Preprocessing query
    query_bersih = preprocessing(query)

    # Mengubah query menjadi vektor TF-IDF
    query_vector = tfidf_vectorizer.transform([query_bersih])

    # Menghitung Cosine Similarity
    skor = cosine_similarity(
        query_vector,
        tfidf_matrix
    ).flatten()

    # Mengambil 5 dokumen dengan skor tertinggi
    indeks_teratas = skor.argsort()[::-1][:top_k]

    # Mengambil ID dokumen hasil pencarian
    dokumen_ditemukan = [
        df.iloc[i]["id_dokumen"]
        for i in indeks_teratas
    ]

    # Menghitung jumlah dokumen relevan yang ditemukan
    relevan_ditemukan = len(
        set(dokumen_ditemukan) & ground_truth[kode_query]
    )

    # Menghitung Precision dan Recall
    precision = relevan_ditemukan / top_k

    recall = (
        relevan_ditemukan / len(ground_truth[kode_query])
    )

    # Menghitung F1-Score
    if precision + recall > 0:
        f1_score = (
            2 * precision * recall / (precision + recall)
        )
    else:
        f1_score = 0

    hasil_evaluasi.append({
        "Query": kode_query,
        "Kata Kunci": query,
        "Dokumen Ditemukan": ", ".join(dokumen_ditemukan),
        "Relevan Ditemukan": relevan_ditemukan,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1_score
    })

# Menampilkan hasil evaluasi
df_evaluasi = pd.DataFrame(hasil_evaluasi)

display(df_evaluasi.round(3))
```

```python
# Menghitung rata-rata setiap metrik
rata_rata = df_evaluasi[
    ["Precision", "Recall", "F1-Score"]
].mean()

print("RATA-RATA HASIL EVALUASI DOCUFIND")
print("--------------------------------")

print("Precision :", round(rata_rata["Precision"], 3))
print("Recall    :", round(rata_rata["Recall"], 3))
print("F1-Score  :", round(rata_rata["F1-Score"], 3))
```
